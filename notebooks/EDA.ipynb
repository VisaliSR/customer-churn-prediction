```python
import numpy as np
df["TotalCharges"]=df["TotalCharges"].replace(" ",np.nan)
#replace empty string with Nan value
```

```python
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"])
df["TotalCharges"].dtype
#Converting the Nan value
```

```python
df["TotalCharges"].isnull().sum()
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
df["TotalCharges"].isnull().sum()
#The missing TotalCharges values corresponded to customers with zero tenure.
# Since these customers had just joined the service and had not yet accumulated charges,
# I replaced the missing values with 0 rather than using statistical imputation or removing the records.
```

```python
columns = ["tenure", "MonthlyCharges", "TotalCharges", "Contract","PaymentMethod", "InternetService", "TechSupport", "Churn"]
df[columns].sample(10)  # Displays 5 random rows as a table
```

```python
median_churned_tenure = df[df["Churn"] == "Yes"]["tenure"].median()
print(f"tenure median for churned customers {median_churned_tenure}")
median_non_churned_tenure = df[df["Churn"]=="No"]["tenure"].median()
print(f"tenure median for non churned customers {median_non_churned_tenure}")
```

```python
df["churn_num"]=df["Churn"].map({"Yes":1,"No":0})
pivot_table=df.pivot_table(values="churn_num",index="Contract",aggfunc="mean")*100
pivot_table.columns=["Churned_percent"]
pivot_table["Non_churned_percent"]=100-pivot_table["Churned_percent"]
print(pivot_table)
```

```python
contract_tenure_pivot=df.pivot_table(values="tenure",index="Contract",aggfunc="median")
contract_tenure_pivot.columns=["Tenure"]
print(contract_tenure_pivot)
```

```python
med_ten=df.groupby("Contract")["tenure"].median()
print(med_ten)
```

```python
median_monthlycharge_churned=df[df["Churn"]=="Yes"]["MonthlyCharges"].median()
print(f"median for monthlycharge churned {median_monthlycharge_churned}")
median_monthlycharge_retained=df[df["Churn"]=="No"]["MonthlyCharges"].median()
print(f"median for monthlycharge retained {median_monthlycharge_retained}")
```

```python
tech_support=df.groupby("TechSupport")["churn_num"].mean()*100
print(tech_support)
```

```python
online_security=df.groupby("OnlineSecurity")["churn_num"].mean()*100
print(online_security)
```

```python
churned_internet_percentage=df.groupby("InternetService")["churn_num"].mean()*100
print(churned_internet_percentage)
```

```python
fiber_techsupport=df[df["InternetService"]=="Fiber optic"].pivot_table(
    values="churn_num",index=["InternetService","TechSupport"],aggfunc="mean"
)*100
fiber_techsupport.columns=["churned"]
fiber_techsupport["retained"]=100-fiber_techsupport["churned"]
print(fiber_techsupport)
```

```python
# Key Findings

# 1. Month-to-month customers are the highest-risk segment.

# Customers on month-to-month contracts have a churn rate of 42.7%, compared to 11.3% for one-year contracts and 2.8% for two-year contracts.
# Retention efforts should primarily focus on month-to-month customers.

# 2. Customer churn is concentrated in the early stages of the customer lifecycle.

# The median tenure of churned customers is 10 months, compared to 38 months for retained customers.
# This suggests that the first year is a critical period for customer retention.

# 3. Technical support is strongly associated with customer retention.

# Customers without TechSupport have a churn rate of 41.6%, while customers with TechSupport have a churn rate of 15.2%.
# Improving support quality and increasing adoption of support services may help reduce churn.

# 4. Online security shows a similar pattern to technical support.

# Customers without OnlineSecurity have a churn rate of 41.8%, compared to 14.6% for customers who have it.
# Security-related services appear to be an important part of perceived customer value.

# 5. Fiber optic customers without technical support represent a particularly high-risk group.

# Fiber optic customers without TechSupport have a churn rate of 49.4%, compared to 22.6% for Fiber customers with TechSupport.
# This segment should be prioritized for retention initiatives.
```
